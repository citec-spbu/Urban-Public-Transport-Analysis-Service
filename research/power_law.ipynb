## Использование
Добавьте кнопку `Переключить просмотр` на страницу, отображаемую `CityMap.jsx`, нажмите на кнопку, чтобы переключиться на страницу отображения мощности. Нажмите кнопку `Посмотреть Законы власти`, чтобы увидеть информацию о мощности графа запросов. Нажмите на кнопку `Вернуться к карте`, чтобы вернуться к отображению диаграммы общественного транспорта.

Пользователь выбирает метод запроса на экране `Home`, вводит условия для перехода на страницу `CityMap`, а функция `handleSendDataButtonClick()` отправляет информацию об узлах и ребрах текущей карты непосредственно на бэкэнд для обработки, поэтому нет необходимости вводить информацию для фильтрации повторно.


## Метод оценки
Имеется четыре данных для информации об узлах общественного транспорта `center_count`, `closeness_centrality`, `betweenness_centrality`, `pagerank`, и после многократной проверки и опроса смежных знаний был сделан вывод, что использование `pagerank` для подгонки является наилучшим.

Определяется простая функция распределения по закону мощности `power_law(x, alpha)`, где x - независимая переменная, а alpha - индекс закона мощности. Распределение по закону мощности подгоняется с помощью функции `scipy.optimize.curve_fit`. Функция `curve_fit` используется для поиска оптимального значения alpha, которое позволяет функции закона мощности наилучшим образом описывать предоставленные данные PageRank.

## Анализ вида
Горизонтальная координата (значение PageRank): большее значение PageRank указывает на то, что узел является более важным в сети.

Вертикальная координата (кумулятивная функция распределения): большие значения кумулятивной функции распределения указывают на то, что больше узлов имеют значения PageRank меньше или равные текущему значению.

Оранжевая линия (подогнанная CDF распределения по закону мощности): эта линия представляет собой кумулятивную функцию распределения, полученную путем подгонки модели закона мощности. Она показывает тенденцию общего распределения значений PageRank и аппроксимирует это распределение моделью закона мощности.

Синие точки (CDF исходных данных): эти точки представляют собой кумулятивную функцию распределения фактических данных. Каждая точка соответствует значению PageRank и соответствующей ему кумулятивной вероятности.
